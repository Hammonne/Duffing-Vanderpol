# 05 - Numerical-method benchmark

Choice of the integrator for the full (reference) solution. We compare fixed-step RK4 against scipy's RK45, DOP853 and Radau on Duffing, Van der Pol and the combined system. See `docs/benchmark.md` for the full table; **DOP853** is adopted as the default.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import numpy as np
import matplotlib.pyplot as plt
from dvdp import renormalization as rg, integrators as it, perturbation as pert, metrics as mt
from dvdp.systems import full_rhs
from scipy.integrate import solve_ivp
```

```python
t = np.linspace(0, 50, 2001)
ref = solve_ivp(full_rhs,(0,50),[1,0],args=(1.0,1.0,1.0),t_eval=t,method='DOP853',rtol=1e-13,atol=1e-13).y[0]
for m in ['RK45','DOP853','Radau']:
    s = solve_ivp(full_rhs,(0,50),[1,0],args=(1.0,1.0,1.0),t_eval=t,method=m,rtol=1e-10,atol=1e-12)
    print(f'{m:8s} rms={np.linalg.norm(s.y[0]-ref)/np.sqrt(len(ref)):.2e}  nfev={s.nfev}')
```
