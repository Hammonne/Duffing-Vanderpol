# 01 - Perturbation theory vs. the exact solution

We show that the naive first-order perturbative correction to the harmonic oscillator develops **secular growth** (unbounded amplitude) for both the Duffing and Van der Pol deformations, motivating the renormalization procedure. Mirrors Sec. II.A of the proposal.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import numpy as np
import matplotlib.pyplot as plt
from dvdp import renormalization as rg, integrators as it, perturbation as pert, metrics as mt
```

```python
t = np.linspace(0, 80, 3000)
_, phi1_D, _ = pert.solve_first_order(pert.source_duffing, 1.0, t_eval=t)
_, phi1_V, _ = pert.solve_first_order(pert.source_vdp, 1.0, t_eval=t)
plt.figure(figsize=(8,4.5))
plt.plot(t, phi1_D, label='Duffing $\\varphi^{(1)}$')
plt.plot(t, phi1_V, label='Van der Pol $\\varphi^{(1)}$')
plt.xlabel('t'); plt.ylabel('$\\varphi^{(1)}$'); plt.legend(); plt.grid(alpha=.3)
plt.title('Secular growth of the first-order correction'); plt.show()
```

The corrections grow without bound: adding `eps*phi1` to the background cannot approximate the full solution for long times. Renormalization reabsorbs this growth into slowly varying amplitude/phase.
