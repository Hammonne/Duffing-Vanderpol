# 02 - Duffing renormalization (frequency shift)

Duffing deforms the oscillator with a cubic term. The renormalization reabsorbs the secular growth into a **frequency shift** $\beta_\theta(A)=\tfrac{3}{8}A^2$ (Bender-Orszag), leaving the amplitude unchanged ($\beta_A\approx0$). Mirrors Sec. II.B / Figs. 1, 3.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import numpy as np
import matplotlib.pyplot as plt
from dvdp import renormalization as rg, integrators as it, perturbation as pert, metrics as mt
```

```python
A = np.linspace(0.05, 3.0, 40)
bA, bth = rg.extract_generators('duffing', A)
plt.figure(figsize=(7,4.5))
plt.plot(A, bth, 'C3', label=r'$\beta_\theta(A)$ (numerical)')
plt.plot(A, 0.375*A**2, 'k--', label=r'$\frac38 A^2$ (Bender-Orszag)')
plt.plot(A, bA, 'C0', label=r'$\beta_A(A)\approx0$')
plt.xlabel('A'); plt.legend(); plt.grid(alpha=.3)
plt.title('Duffing generators'); plt.show()
print('beta_theta/A^2 at A=1:', np.interp(1.0, A, bth))
```

```python
t = np.linspace(0, 100, 4000); eps = 0.1
pf,_ = it.solve_full(t, eps=eps, A0=1.0)
pn,_ = it.solve_naive(t, eps=eps, A0=1.0)
res = rg.renormalize(t, eps=eps, A0=1.0)
plt.figure(figsize=(9,4.5))
plt.plot(t, pn, 'k--', lw=1, label='naive')
plt.plot(t, pf, 'C0', lw=1.3, label='full')
plt.plot(t, res.phi_R, 'C3', lw=1.3, label='renormalized')
plt.ylim(-4,4); plt.xlabel('t'); plt.ylabel('$\\varphi$'); plt.legend(); plt.grid(alpha=.3)
plt.title(f'Duffing eps={eps}'); plt.show()
print('RMS  RG=%.4f  naive=%.4f' % (mt.rms_error(res.phi_R,pf), mt.rms_error(pn,pf)))
```
