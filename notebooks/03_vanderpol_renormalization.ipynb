# 03 - Van der Pol renormalization (amplitude flow)

Van der Pol requires an **amplitude** flow $\beta_A(A)=\tfrac12A(1-A^2/4)$ (zero at the limit cycle $A=2$), with negligible frequency shift. Mirrors Sec. II.C / Fig. 4.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import numpy as np
import matplotlib.pyplot as plt
from dvdp import renormalization as rg, integrators as it, perturbation as pert, metrics as mt
```

```python
A = np.linspace(0.05, 3.0, 40)
bA, bth = rg.extract_generators('vdp', A)
plt.figure(figsize=(7,4.5))
plt.plot(A, bA, 'C0', label=r'$\beta_A(A)$ (numerical)')
plt.plot(A, 0.5*A*(1-A**2/4), 'k--', label=r'$\frac12A(1-A^2/4)$')
plt.axhline(0, color='gray', lw=.5); plt.axvline(2, color='gray', ls=':', lw=.8)
plt.xlabel('A'); plt.legend(); plt.grid(alpha=.3); plt.title('Van der Pol generators'); plt.show()
```

```python
t = np.linspace(0, 100, 4000); alpha = 0.1
pf,qf = it.solve_full(t, alpha=alpha, A0=1.0)
res = rg.renormalize(t, alpha=alpha, A0=1.0)
fig,(a1,a2)=plt.subplots(1,2,figsize=(13,5))
a1.plot(t, pf,'C0',label='full'); a1.plot(t,res.phi_R,'C3',label='renorm'); a1.legend(); a1.grid(alpha=.3); a1.set_xlabel('t'); a1.set_title(f'VdP alpha={alpha}')
a2.plot(pf,qf,'C0',lw=1,label='full'); a2.plot(res.phi_R,res.pi_R,'C3',label='renorm'); a2.axis('equal'); a2.legend(); a2.grid(alpha=.3); a2.set_title('phase space (limit cycle)')
plt.show(); print('A final =', res.A_t[-1])
```
