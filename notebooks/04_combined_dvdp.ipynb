# 04 - Combined Duffing-Van der Pol (the universal method)

The master equation

$$\dot\varphi=\Pi,\quad \dot\Pi=-\omega^2\varphi-\epsilon\varphi^3+\alpha(1-\varphi^2)\Pi$$

is renormalized with a **single universal engine**. Because the first-order equation is linear, the generators superpose:
$\beta=\epsilon\,\beta^{\rm Duf}+\alpha\,\beta^{\rm VdP}$. Duffing contributes a frequency shift, Van der Pol an amplitude flow; both act simultaneously.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import numpy as np
import matplotlib.pyplot as plt
from dvdp import renormalization as rg, integrators as it, perturbation as pert, metrics as mt
```

```python
def show(eps, alpha, tmax=100):
    t = np.linspace(0, tmax, 4000)
    pf,qf = it.solve_full(t, eps=eps, alpha=alpha, A0=1.0)
    pn,_  = it.solve_naive(t, eps=eps, alpha=alpha, A0=1.0)
    res   = rg.renormalize(t, eps=eps, alpha=alpha, A0=1.0)
    fig,(a1,a2)=plt.subplots(1,2,figsize=(13,5))
    a1.plot(t,pn,'k--',lw=.8,alpha=.6,label='naive'); a1.plot(t,pf,'C0',lw=1.2,label='full'); a1.plot(t,res.phi_R,'C3',lw=1.2,label='renorm')
    a1.set_ylim(-3.2,3.2); a1.legend(); a1.grid(alpha=.3); a1.set_xlabel('t'); a1.set_title(f'DVdP eps={eps}, alpha={alpha}')
    a2.plot(pf,qf,'C0',lw=1,label='full'); a2.plot(res.phi_R,res.pi_R,'C3',label='renorm'); a2.axis('equal'); a2.legend(); a2.grid(alpha=.3); a2.set_title('phase space')
    plt.show(); print('RMS  RG=%.4f  naive=%.4f' % (mt.rms_error(res.phi_R,pf), mt.rms_error(pn,pf)))

show(0.1, 0.1)   # soft coupling
show(1.0, 1.0)   # strong coupling
```

`eps=0` recovers pure Van der Pol and `alpha=0` recovers pure Duffing, so this one engine solves all three problems. For soft coupling the approximation is excellent; for strong coupling it captures the amplitude envelope and limit-cycle size, with residual phase drift that is the expected first-order limitation (the true strong-coupling limit cycle is anharmonic).
